```python
# Policy Iteration in Deterministic Grid World

# 1. Define MDP

states = ['S0', 'S1', 'G']

actions = ['Down', 'Right', 'Up', 'Left']

gamma = 0.9
theta = 0.0001

normal_reward = -1
goal_reward = 100

# Transition function
# S0 Down -> S1
# S1 Right -> G
# Other actions stay in the same state

def transition(state, action):

    if state == 'S0' and action == 'Down':
        return 'S1'

    elif state == 'S1' and action == 'Right':
        return 'G'

    else:
        return state


# Reward function

def reward(state, action, next_state):

    if next_state == 'G':
        return goal_reward

    return normal_reward


# 2. Initialize policy and state values

policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("Initial Policy:", policy)
print("Initial Values:", V)


# 3. Policy Evaluation

def policy_evaluation(policy, V):

    iteration = 0

    while True:

        delta = 0

        old_values = V.copy()

        for state in ['S0', 'S1']:

            action = policy[state]

            next_state = transition(state, action)

            r = reward(state, action, next_state)

            new_value = r + gamma * old_values[next_state]

            delta = max(delta, abs(new_value - V[state]))

            V[state] = new_value

        iteration += 1

        print("Evaluation", iteration,
              "S0 =", round(V['S0'], 4),
              "S1 =", round(V['S1'], 4))

        if delta < theta:
            break

    return V


# 4. Calculate action values

def calculate_q(state, action, V):

    next_state = transition(state, action)

    r = reward(state, action, next_state)

    return r + gamma * V[next_state]


# 5. Policy Improvement

def policy_improvement(policy, V):

    policy_stable = True

    for state in ['S0', 'S1']:

        old_action = policy[state]

        q_values = {}

        for action in actions:

            q_values[action] = calculate_q(
                state, action, V
            )

        best_action = max(q_values, key=q_values.get)

        policy[state] = best_action

        if old_action != best_action:
            policy_stable = False

    return policy, policy_stable


# 6. Policy Iteration

iteration = 0

while True:

    iteration += 1

    print("\nPolicy Iteration:", iteration)

    V = policy_evaluation(policy, V)

    policy, stable = policy_improvement(policy, V)

    print("Improved Policy:", policy)

    if stable:
        break


# 7. Final State Values

print("\nFinal Optimal State Values")

for state in states:

    print(state, "=", round(V[state], 4))


# 8. Final Action Value Table

print("\nFinal Action Value Table")

for state in ['S0', 'S1']:

    print("\nState:", state)

    for action in actions:

        q = calculate_q(state, action, V)

        print(action, "=", round(q, 4))


# 9. Optimal Policy

print("\nOptimal Policy")

for state in ['S0', 'S1']:

    print(state, "->", policy[state])


# 10. Bellman Optimality Verification

print("\nBellman Optimality Verification")

for state in ['S0', 'S1']:

    q_values = [
        calculate_q(state, action, V)
        for action in actions
    ]

    max_q = max(q_values)

    print(
        state,
        "V*(s) =", round(V[state], 4),
        "Max Q*(s,a) =", round(max_q, 4)
    )

    assert abs(V[state] - max_q) < theta

print("\nBellman Optimality Verified Successfully!")
```
