```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10

# Part B – Initialize tables
N = [0, 0, 0]          # Number of times each machine is played
R = [0, 0, 0]          # Total reward
Q = [0.0, 0.0, 0.0]    # Q-values
UCB = [float('inf')] * M

# ---------------------------------------
# INPUT REWARD PATTERNS
# ---------------------------------------

reward_patterns = []

print("Enter reward pattern for each machine")
print("Example: 3 4 2 5\n")

for machine in machines:
    rewards = list(map(float, input(
        f"Enter rewards for Machine {machine}: "
    ).split()))

    reward_patterns.append(rewards)

print("\nReward Patterns:")
for i in range(M):
    print(machines[i], ":", reward_patterns[i])


# Keeps track of which reward comes next
reward_index = [0, 0, 0]

total_reward = 0


# ---------------------------------------
# UCB ALGORITHM
# ---------------------------------------

for play in range(1, N_play + 1):

    print("\n======================================")
    print("Play", play)
    print("======================================")

    # Calculate UCB
    for i in range(M):

        if N[i] == 0:
            # Give unplayed machines priority
            UCB[i] = float('inf')

        else:
            UCB[i] = Q[i] + math.sqrt(
                (2 * math.log(play)) / N[i]
            )

    # Select machine with highest UCB
    selected = UCB.index(max(UCB))

    # Display UCB values
    print("\nUCB Values:")

    for i in range(M):

        if UCB[i] == float('inf'):
            print(machines[i], "= Infinity")
        else:
            print(
                machines[i],
                "=",
                round(UCB[i], 4)
            )

    print("\nSelected Machine:", machines[selected])

    # ---------------------------------------
    # GET REWARD FROM PATTERN
    # ---------------------------------------

    # Use modulo so that rewards repeat
    index = reward_index[selected] % len(
        reward_patterns[selected]
    )

    reward = reward_patterns[selected][index]

    print("Reward obtained:", reward)

    # Move to next reward
    reward_index[selected] += 1

    # ---------------------------------------
    # UPDATE N
    # ---------------------------------------

    N[selected] += 1

    # ---------------------------------------
    # UPDATE R
    # ---------------------------------------

    R[selected] += reward

    # ---------------------------------------
    # UPDATE Q
    # ---------------------------------------

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    # Update total reward
    total_reward += reward

    print("Number of times played:", N[selected])
    print("Updated Q-value:", round(Q[selected], 4))


# ---------------------------------------
# FINAL RESULTS
# ---------------------------------------

print("\n\n======================================")
print("           FINAL RESULTS")
print("======================================")

# Calculate final UCB
for i in range(M):

    if N[i] == 0:
        UCB[i] = float('inf')

    else:
        UCB[i] = Q[i] + math.sqrt(
            (2 * math.log(N_play)) / N[i]
        )


print("\nMachine\tN\tR\tUCB\t\tQ")
print("----------------------------------------------")

for i in range(M):

    if UCB[i] == float('inf'):
        ucb_value = "Infinity"

    else:
        ucb_value = round(UCB[i], 4)

    print(
        machines[i],
        "\t",
        N[i],
        "\t",
        R[i],
        "\t",
        ucb_value,
        "\t\t",
        round(Q[i], 4)
    )

print("\n======================================")
print("Total Reward =", total_reward)
print("======================================")
```
