```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']
N_play = 10
epsilon = 0.2
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
total_reward = 0
print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("------------------------------------------")
for play in range(1, N_play + 1):
    r = random.random()
    best_machine = max(Q, key=Q.get)
    if r < epsilon:
        other_machines = [m for m in machines if m != best_machine]
        selected_machine = random.choice(other_machines)
        print(f"\nPlay {play}: Exploration")
    else:
        selected_machine = best_machine
        print(f"\nPlay {play}: Exploitation")
    print("Selected Machine:", selected_machine)
    reward = float(input(f"Enter reward for machine {selected_machine}: "))
    N[selected_machine] += 1
    R[selected_machine].append(reward)
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )
    total_reward += reward
    print("Updated Q-values:", Q)
# Part F - Display Results
print("\n========== FINAL RESULTS ==========")
print("\nN Table - Participation Count")
for machine in machines:
    print(machine, ":", N[machine])
print("\nR Table - Rewards")
for machine in machines:
    print(machine, ":", R[machine])
print("\nQ Table - Estimated Values")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))
print("\nTotal Reward Received:", total_reward)
```
