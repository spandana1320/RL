```python
# Greedy Multi-Armed Bandit
# Incremental Update Method

# Number of machines
n = int(input("Enter number of machines: "))

Q = {}
N = {}
rewards = {}

# Task 1: Initialize
for i in range(n):
    machine = input("Enter machine name: ")
    reward_list = list(map(int, input("Enter rewards: ").split()))

    rewards[machine] = reward_list
    Q[machine] = 0
    N[machine] = 0

print("\n--- Incremental Updates ---")

# Task 2: Update after every play
for machine in rewards:

    print(f"\nMachine {machine}")

    for R in rewards[machine]:

        # Increment number of plays
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (R - Q[machine]) / N[machine]

        print(
            f"Reward = {R}, "
            f"N({machine}) = {N[machine]}, "
            f"Q({machine}) = {Q[machine]:.2f}"
        )

# Task 3: Display final Q values
print("\n--- Final Estimated Values ---")

for machine in Q:
    print(f"Machine {machine}")
    print(f"Rewards: {rewards[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

# Greedy selection
selected_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", selected_machine)
```
