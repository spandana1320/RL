```python
# Assignment 8: Calculate State Value and Action Value
# Reinforcement Learning - Week 6

# Discount factor
gamma = 0.9

# Grid World states
states = ["S0", "S1", "G", "X", "S3", "S4"]

# Rewards
normal_reward = -1
goal_reward = 10
obstacle_reward = -5

# Policy given in the assignment
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# State transitions
# Format: (next_state, reward, terminal)
transitions = {
    ("S0", "Right"): ("S1", normal_reward, False),
    ("S0", "Up"): ("S0", normal_reward, False),
    ("S0", "Down"): ("S0", normal_reward, False),
    ("S0", "Left"): ("S0", normal_reward, False),

    ("S1", "Right"): ("G", goal_reward, True),
    ("S1", "Down"): ("S3", normal_reward, False),
    ("S1", "Up"): ("S1", normal_reward, False),
    ("S1", "Left"): ("S0", normal_reward, False),

    ("S3", "Right"): ("S4", normal_reward, False),
    ("S3", "Left"): ("X", obstacle_reward, True),
    ("S3", "Up"): ("S1", normal_reward, False),
    ("S3", "Down"): ("S3", normal_reward, False),

    ("S4", "Up"): ("G", goal_reward, True),
    ("S4", "Down"): ("S4", normal_reward, False),
    ("S4", "Left"): ("S3", normal_reward, False),
    ("S4", "Right"): ("S4", normal_reward, False)
}


# Calculate State Value by following the policy
def state_value_function(start_state):
    current_state = start_state
    total_return = 0
    discount = 1
    step = 1

    print(f"\nCalculating State Value V({start_state})")

    while current_state not in ["G", "X"]:
        action = policy[current_state]
        next_state, reward, terminal = transitions[(current_state, action)]

        contribution = discount * reward
        total_return += contribution

        print(
            f"Step {step}: {current_state} --{action}--> "
            f"{next_state}, Reward = {reward}, "
            f"Discounted reward = {contribution:.3f}"
        )

        if terminal:
            break

        current_state = next_state
        discount *= gamma
        step += 1

    print(f"State Value V({start_state}) = {total_return:.3f}")
    return total_return


# Calculate Action Value by taking the specified action first,
# then following the policy
def action_value_function(state, action):
    next_state, reward, terminal = transitions[(state, action)]

    print(f"\nCalculating Action Value Q({state}, {action})")
    print(f"First action: {state} --{action}--> {next_state}, Reward = {reward}")

    if terminal:
        total_return = reward
    else:
        # Calculate the return after the first action
        following_return = state_value_function(next_state)
        total_return = reward + gamma * following_return

    print(f"Action Value Q({state}, {action}) = {total_return:.3f}")
    return total_return


# Main program
if __name__ == "__main__":
    # Calculate state value for S0
    v_s0 = state_value_function("S0")

    # Calculate action value for S0 and action Up
    q_s0_up = action_value_function("S0", "Up")

    print("\n========== FINAL RESULTS ==========")
    print(f"V(S0) = {v_s0:.3f}")
    print(f"Q(S0, Up) = {q_s0_up:.3f}")
```
