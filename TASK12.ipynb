```python

# Value Iteration in a Deterministic 2x2 Grid World

gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

# Transition rules
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

# Reward function
def get_reward(next_state):
    if next_state == 'G':
        return 100
    return -1

# Value Iteration
iteration = 0

print("Iteration-wise State Values")

while True:
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]

            reward = get_reward(next_state)

            q = reward + gamma * V[next_state]

            q_values.append(q)

        new_V[state] = max(q_values)

        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print("Iteration", iteration, ":", V)

    if delta < theta:
        break

# Calculate final Q-values and optimal policy
print("\nFinal State Values:")

for state in states:
    print(state, "=", V[state])

print("\nFinal Q-Values:")

for state in ['S0', 'S1']:
    q_values = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)

        q_values[action] = reward + gamma * V[next_state]

    print("\nState:", state)

    for action, q in q_values.items():
        print(action, ":", round(q, 2))

    best_action = max(q_values, key=q_values.get)

    print("Optimal Action:", best_action)
```
